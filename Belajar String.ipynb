String Formatting

String formatting memungkinkan Anda membuat string yang memasukkan nilai variabel, yang penting untuk menghasilkan laporan, membuat nama file, membangun query, dan menampilkan informasi geografis. Python menyediakan beberapa metode formatting yang powerful yang memudahkan pembuatan output teks yang terstruktur dengan baik.

F-String Formatting (Direkomendasikan)

F-strings (formatted string literals) adalah cara paling modern dan mudah dibaca untuk memformat string dalam Python. Mereka sangat berguna untuk membuat laporan geografis dan menampilkan informasi koordinat. Untuk menggunakan f-strings, Anda dapat menggunakan prefix f sebelum string, dan kemu dian menggunakan kurung kurawal untuk menyisipkan nilai variabel. Contoh di bawah ini menunjukkan cara menggunakan f-strings untuk memformat string dengan nilai variabel:

```python
location = "Mount Everest"
latitude = 27.9881
longitude = 86.9250
elevation = 8848

#tunggal
location_info = f"Location is {location}"
print(location_info)

#beberapa variabel
Coordinates_info = f'Coordinates: ({latitude},{longitude})'
print(Coordinates_info)

#ringkasan lengkap
summary = f"{location} is located at ({longitude}),({longitude}) with an elevation of {elevation} Meters"
print(summary)
```

Memformat Angka dalam String

Saat bekerja dengan data geografis, Anda sering perlu mengontrol bagaimana angka ditampilkan. Misal-
nya, Anda mungkin ingin menampilkan koordinat dengan jumlah tempat desimal tertentu, atau angka
dengan jumlah digit signifikan tertentu. Untuk melakukan ini, Anda dapat menggunakan f-strings untuk
memformat angka. Spesifier format adalah titik dua : diikuti oleh jumlah tempat desimal atau digit
signifikan. Contoh di bawah ini menunjukkan cara memformat angka dengan jumlah tempat desimal
tertentu:

```python
# Mengontrol tempat desimal untuk koordinat
precise_lat = 40.7771
precise_lon = -74.9999

# Pembulatan ke tempat desimal yang berbeda
coords_2_places = f"coordinates: ({precise_lat:.2f}, ({precise_lon:.2f})"
coords_4_places = f"coordinates: ({precise_lat :.4f}, ({precise_lon:.4f})"

print(coords_2_places)
print(coords_4_places)
```

```python
# Menambahkan separator ribuan untuk angka besar
population = 87732173
area_sqkm = 883.3

formatted_stats = f"NYC Population: {population:,} people, Area {area_sqkm:.1f} km2"
print(formatted_stats)
```

Metode Formatting Legacy

Meskipun f-strings lebih disukai, Anda harus familiar dengan metode formatting lama yang mungkin
Anda temui dalam kode lama. Contoh di bawah ini menunjukkan cara memformat angka dengan jumlah
tempat desimal tertentu menggunakan metode format() . Alih-alih menggunakan kurung kurawal,
Anda menggunakan kurung kurawal dengan angka di dalam untuk menunjukkan posisi variabel yang
akan diformat. Contoh di bawah ini menunjukkan cara memformat angka dengan jumlah tempat desimal
tertentu:

```python
Location = "San Francisco"
lat = 37.7774
lon = 45.888

# Metode format dasar
formatted_1 = "Lokasi {} berada di koordinat ({},{})".format(Location, lat, lon)
print(formatted_1)

# Dengan argumen posisional
formatted_2 = "Lokasi {0} berada di koordinat ({1},{2})".format(Location, lat, lon)
print(formatted_2)

# Dengan argumen bernama
formatted_3 = "lokasi {nama} berada di koordinat ({lat},{lon})".format(nama = location, lat = lat, lon = lon)
print(formatted_3)
```

Contoh Formatting Praktis

Berikut beberapa skenario formatting geospasial umum. Misalnya, Anda mungkin ingin menyimpan
koordinat sebagai string dengan jumlah tempat desimal tertentu dalam file dengan timestamp di nama
file. Contoh di bawah ini menunjukkan cara melakukan ini:

```python
# Membuat nama file dengan timestamp dan koordinat
from datetime import datetime, date
current_time = datetime.now()
survey_lat = 45.33321
survey_lon = -121.17262

filename = f"survey_{current_time.strftime('%y%m%d')}_{survey_lat:.4f}N_{abs(survey_lon):4f}W.csv"
print(filename)

# Membuat representasi Well-Known Text (WKT)
wkt_point = f"POINT {survey_lon},{survey_lat}"
print(wkt_point)
```
