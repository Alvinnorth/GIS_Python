Sintaks For Loop Dasar
Struktur dasar for loop melibatkan penentuan variabel untuk mewakili setiap item dalam urutan, diikuti
oleh urutan yang ingin Anda iterasi:

```python
coordinates = [(35.7776, 13.8887),
               (57.8887, 13.8888),
               (61.7767, 13.8887),
               ]
for lat, lon in coordinates:
    print(f"Latitude: {lat}, Longitude: {lon}")


#Perhatikan bagaimana kita menggunakan tuple unpacking di sini—setiap tuple koordinat (lat, lon) secara otomatis di-unpack menjadi variabel lat dan lon yang terpisah. Ini adalah pola yang sangat umum dan berguna saat bekerja dengan data koordinat.
```

```python
nama_kandidat = ["Alvin", "Ulum", "Faza"]

for kandidat in nama_kandidat:
    print(f"Orang ini yang akan sukses : {nama_kandidat}")

```

```python
for a, kandidat in enumerate(nama_kandidat, start = 1):
    print(f"{a}. {kandidat}")
```

Memproses Beberapa Titik Data
For loop menjadi sangat powerful ketika Anda perlu melakukan perhitungan atau transformasi pada
beberapa titik data. Berikut cara Anda dapat menggunakan loop untuk menghitung jarak dari titik
referensi:

```python
def calculate_distance(lat1,lon1, lat2, lon2):
    # Placeholder untuk logika perhitungan jarak
    return((lat2 - lat1)**2 + (lon2 - lon1)**2)**0.5

reference_point = (0,0) # Titik referensi (lintang, bujur)
for lat, lon in coordinates:
    distance = calculate_distance(reference_point[0], reference_point[1], lat, lon)
    print(f'Distance from {reference_point} to ({lat},{lon}):{distance:.2f}')
```

Melakukan Iterasi Melalui Koleksi Geografis
For loop bekerja dengan banyak jenis koleksi yang berbeda. Berikut beberapa pola umum yang akan Anda
temui dalam pekerjaan geospasial:

```python
cities = ["Bogor", "Depok", "Jakarta", "Bekasi"]

print("Major Cities in Indonesia:")
for a, city in enumerate(cities):
    print(f"{a+1}. {city}")

```

```python
Pola_Ruang = ["Hutan", "Pemukiman", "Badan Air", "Perjas"]
print("Jenis Pola Ruang:")
for a, polaruang in enumerate(Pola_Ruang):
    print(f'{a+1}. {polaruang}')

#Dalam contoh di atas, kita menggunakan for loop untuk melakukan iterasi melalui list nama kota.
#Fungsi enumerate digunakan untuk mendapatkan indeks setiap item dalam list. Ini memungkinkan kita
#mencetak indeks dan nama kota dalam satu baris. Indeks dimulai dari 0 dan bertambah 1 untuk setiap
#item dalam list. Ini dapat berguna ketika Anda perlu melacak posisi setiap item dalam list.
```

```python
for a, pola_ruang in enumerate(Pola_Ruang, start = 1):
    print(f"{a}.{pola_ruang}")
```

```python
tuplah = ["Lahan Terbuka", "Ladang", "Pertanian", "Permukiman", "Sungai", "Danau"]
for a, landuse in enumerate(tuplah, start = 1):
    print(f"{a}. {landuse}")
```

```python
City_name = ["Bogor", "Depok", "Tangerang", "Jakarta"]
print("Kota Sekitar:")
for a, namakota in enumerate(City_name, start = 1):
    print((f"{a}.{namakota}"))
```

Struktur While Loop Dasar
While loop memerlukan kondisi yang dapat dievaluasi sebagai True atau False . Loop berlanjut selama
kondisi adalah True dan berhenti ketika menjadi False :

```python
counter = 0 
while counter <len(coordinates):
    lat, lon = coordinates[counter]
    print(f"processing coordinate:({lat},{lon})")
    counter +=1

#Catatan Penting: Saat menggunakan while loop, selalu pastikan kondisi Anda pada akhirnya akan
#menjadi False , atau Anda akan membuat infinite loop! Dalam contoh di atas, kita menambah variabel
#counter di setiap iterasi untuk memastikan loop pada akhirnya akan berakhir.
```

```python
buffer = [300, 500, 700, 1300]
batas = 0
while batas <max (buffer):
    print(f"Buffer yang dicari:{buffer}")
    batas +=500
```

```python
nilai = [65, 80, 95, 100, 50, 120]
batas = 0
while batas <max (nilai):
    print(f"Nilai yang dicari:{nilai}")
    batas +=300
```
