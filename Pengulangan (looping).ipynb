Sintaks For Loop Dasar
Struktur dasar for loop melibatkan penentuan variabel untuk mewakili setiap item dalam urutan, diikuti
oleh urutan yang ingin Anda iterasi:

```python
coordinates = [(35.7776, 13.8887),
               (57.8887, 13.8888),
               (61.7767, 13.8887),
               ]
for lat, lon in coordinates:
    print(f"Latitude: {lat}, Longitude: {lon}")


#Perhatikan bagaimana kita menggunakan tuple unpacking di sini—setiap tuple koordinat (lat, lon) secara otomatis di-unpack menjadi variabel lat dan lon yang terpisah. Ini adalah pola yang sangat umum dan berguna saat bekerja dengan data koordinat.
```

Memproses Beberapa Titik Data
For loop menjadi sangat powerful ketika Anda perlu melakukan perhitungan atau transformasi pada
beberapa titik data. Berikut cara Anda dapat menggunakan loop untuk menghitung jarak dari titik
referensi:

```python
def calculate_distance(lat1,lon1, lat2, lon2):
    # Placeholder untuk logika perhitungan jarak
    return((lat2 - lat1)**2 + (lon2 - lon1)**2)**0.5

reference_point = (0,0) # Titik referensi (lintang, bujur)
for lat, lon in coordinates:
    distance = calculate_distance(reference_point[0], reference_point[1], lat, lon)
    print(f'Distance from {reference_point} to ({lat},{lon}):{distance:.2f}')
```
