 Latihan 1: Memanipulasi String Lokasi Geografis
• Buat string yang merepresentasikan nama fitur geografis (misalnya, "Amazon River" ).
• Konversi string ke lowercase dan kemudian ke uppercase.
• Gabungkan string dengan nama negara (misalnya, "Brazil" ) untuk membuat nama lokasi lengkap.
• Ulangi string tiga kali, pisahkan setiap pengulangan dengan tanda hubung ( - ).

```python
Nama_Negara = "Brazil"
Sungai = "Amazon River"
sungai_lowecase = Sungai.lower()
sungai_uppercase = Sungai.upper()
print(sungai_lowecase)
print(sungai_uppercase)
```

```python
Combination = Nama_Negara + " Have " + Sungai
print(Combination)
Pengulangan = " - ".join([Combination] * 30)
print(Pengulangan)
```

 Latihan 2: Mengekstrak dan Memformat Koordinat
• Diberikan string dengan format "latitude, longitude" (misalnya, "40.7128N, 74.0060W" ), ekstrak nilai numerik lintang dan bujur.
• Konversi nilai ini ke float dan hapus indikator arah ( N , S , E , W ).
• Format koordinat menjadi string WKT POINT (misalnya, "POINT(-74.0060 40.7128)" ).

```python
coordinates_string = "40.7128N, 74.0060W"
replaced_string = coordinates_string.replace("N", "").replace("W", "")
lat_str, lon_str = replaced_string.split(",")
latitude = float(lat_str)
longitude = float(lon_str)
print(f"latitude: {latitude}, longitude: {longitude}")

wkt_point = f"POINT {longitude} {latitude}"
print(wkt_point)
```

Latihan 3: Membangun Query SQL Dinamis
• Diberikan nama tabel dan kondisi, bangun string query SQL secara dinamis.
• Contoh: Jika table_name = "cities" dan condition = "population > 1000000" , query harus
menjadi "SELECT * FROM cities WHERE population > 1000000;" .
• Tambahkan kondisi tambahan secara dinamis, seperti klausa AND .

```python
table_name = "cities"
population = 1000000
region = "Bogor Cities"

sql_querry = f"SELECT * FROM {table_name} WHERE population > {population:,} AND region = '{region}'"
print(sql_querry)
```

Latihan 4: Normalisasi dan Pembersihan String
• Diberikan list nama kota dengan formatting yang tidak konsisten (misalnya,
[" new york ", "Los ANGELES", " CHICAGO"] ), normalisasi nama dengan:
‣ Menghilangkan leading atau trailing whitespace.
‣ Mengkonversinya ke title case (misalnya, "New York" , "Los Angeles" , "Chicago" ).
• Pastikan output adalah list nama kota yang bersih.

```python
City_name = [" new york ", "Los ANGELES", " CHICAGO"]
edit_city_name = [city.strip().title() for city in City_name]
print(edit_city_name)
```

 Latihan 5: Parsing dan Mengekstrak Informasi Alamat
• Diberikan string dalam format "Street, City, Country" (misalnya,
"123 Main St, Springfield, USA" ), tulis fungsi yang parsing string menjadi dictionary dengan
key street , city , dan country .
• Fungsi harus mengembalikan dictionary
seperti {"street": "123 Main St", "city": "Springfield", "country": "USA"} .

```python
cities_data = "123 Main St, Springfield, USA"
parsing_data = cities_data.split(", ")
street, city, country = parsing_data
print(f"Street : {street}")
print(f"City :{city}")
print(f"Country : {country}")
```
